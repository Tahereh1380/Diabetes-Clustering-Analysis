```python
import pandas  as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes.csv")
```

```python
df['AgeGroup'] = pd.cut(df['Age'], bins=[20, 30, 40, 50, 60, 80], labels=["20s", "30s", "40s", "50s", "60+"])
sns.countplot(x='AgeGroup', hue='Outcome', data=df)
plt.title("Diabetes Count Across Age Groups")
plt.show()
```

```python
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"
    
df["BMI_category"] = df["BMI"].apply(categorize_bmi)
bmi_distribution_df = df["BMI_category"].value_counts().reset_index()
bmi_distribution_df.columns = ["BMI_category", "Count"]
bmi_distribution_df
```
